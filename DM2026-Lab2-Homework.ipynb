```python
# from common.py
import csv
import hashlib
import json
import re
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score
from sklearn.model_selection import StratifiedGroupKFold

LABELS = ["anger", "disgust", "fear", "joy", "sadness", "surprise"]
LABEL2ID = {l: i for i, l in enumerate(LABELS)}
N_FOLDS = 5
REPEAT_SEEDS = [0, 1, 2]  # one StratifiedGroupKFold split per seed


def norm_text(t: str) -> str:
    """Key used to find duplicate posts: lowercase, collapse whitespace."""
    return re.sub(r"\s+", " ", t).strip().lower()


def load() -> tuple[pd.DataFrame, pd.DataFrame]:
    """Return (train, test).

    train: post_id, text, label, y   (order of data_identification.csv)
    test:  post_id, text             (order of sampleSubmission.csv)
    """
    posts = pd.DataFrame(json.loads((RAW / "final_posts.json").read_text(encoding="utf-8")))
    ident = pd.read_csv(RAW / "data_identification.csv", dtype=str)
    labels = pd.read_csv(RAW / "emotion.csv", dtype=str)
    sample = pd.read_csv(RAW / "sampleSubmission.csv", dtype=str)

    assert posts["post_id"].is_unique and ident["post_id"].is_unique
    assert set(posts["post_id"]) == set(ident["post_id"])
    df = ident.merge(posts, on="post_id", how="left", validate="1:1")

    train = df[df["set"] == "train"].merge(labels, on="post_id", how="left", validate="1:1")
    assert train["label"].notna().all() and set(train["label"]) == set(LABELS)
    train = train[["post_id", "text", "label"]].reset_index(drop=True)
    train["y"] = train["label"].map(LABEL2ID).astype(int)

    test_ids = df.loc[df["set"] == "test", "post_id"]
    assert set(test_ids) == set(sample["post_id"]) and len(sample) == len(test_ids)
    test = sample[["post_id"]].merge(df[["post_id", "text"]], on="post_id", how="left", validate="1:1")
    assert test["text"].notna().all()
    return train, test


def make_folds(train: pd.DataFrame) -> pd.DataFrame:
    """Repeated StratifiedGroupKFold; duplicate texts share a group, so a
    validation post never has an identical twin in its training folds."""
    groups = train["text"].map(norm_text).factorize()[0]
    out = pd.DataFrame({"post_id": train["post_id"]})
    for r, seed in enumerate(REPEAT_SEEDS):
        f = np.full(len(train), -1)
        skf = StratifiedGroupKFold(n_splits=N_FOLDS, shuffle=True, random_state=seed)
        for k, (_, va) in enumerate(skf.split(train["text"], train["y"], groups)):
            f[va] = k
        assert (f >= 0).all()
        # no group may straddle folds
        g2f = pd.DataFrame({"g": groups, "f": f}).groupby("g")["f"].nunique()
        assert (g2f == 1).all(), "duplicate group split across folds"
        out[f"rep{r}"] = f
    return out


def get_folds(train: pd.DataFrame) -> pd.DataFrame:
    """Load the frozen folds (create them once)."""
    if FOLDS_PATH.exists():
        folds = pd.read_csv(FOLDS_PATH, dtype={"post_id": str})
    else:
        folds = make_folds(train)
        folds.to_csv(FOLDS_PATH, index=False, lineterminator="\n")
    assert folds["post_id"].tolist() == train["post_id"].tolist()
    return folds


def sha256(path: Path) -> str:
    return hashlib.sha256(Path(path).read_bytes()).hexdigest()
```

```python
# from prep.py
import json
from collections import Counter

import pandas as pd

out = {"n_train": len(train), "n_test": len(test)}
out["train_label_counts"] = train["label"].value_counts().reindex(LABELS).to_dict()

# duplicate structure
tr_key = train["text"].map(norm_text)
te_key = test["text"].map(norm_text)
grp = train.assign(k=tr_key).groupby("k")["label"].agg(list)
multi = grp[grp.map(len) >= 2]
out["train_dup_groups"] = int(len(multi))
out["train_dup_posts"] = int(multi.map(len).sum())
out["train_dup_groups_conflicting"] = int((multi.map(lambda v: len(set(v)) > 1)).sum())

# leave-one-out: predict each member of a train duplicate group by the
# majority label of the other members (ties -> counted as wrong)
hit = tot = 0
for labs in multi:
    for i, lab in enumerate(labs):
        rest = Counter(labs[:i] + labs[i + 1:]).most_common()
        tot += 1
        if len(rest) == 1 or rest[0][1] > rest[1][1]:
            hit += int(rest[0][0] == lab)
out["loo_copy_rule_acc"] = hit / tot if tot else None
out["loo_copy_rule_n"] = tot

in_train = te_key.isin(set(tr_key))
out["test_posts_with_train_twin"] = int(in_train.sum())
out["test_twin_groups_label_conflict"] = int(
    sum(len(set(grp[k])) > 1 for k in te_key[in_train].unique())
)
out["test_test_dup_posts"] = int(te_key.duplicated(keep=False).sum())

# adapted
print(json.dumps(out, indent=2, ensure_ascii=False))
```

```python
# from a_tfidf.py
from scipy.sparse import hstack
from sklearn.feature_extraction.text import TfidfVectorizer


def featurize(fit_texts, *apply):
    word = TfidfVectorizer(analyzer="word", ngram_range=(1, 2), sublinear_tf=True,
                           token_pattern=r"(?u)\b\w+\b|[^\w\s]")
    char = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), min_df=2, sublinear_tf=True)
    Xw, Xc = word.fit_transform(fit_texts), char.fit_transform(fit_texts)
    out = [hstack([Xw, Xc]).tocsr()]
    for t in apply:
        out.append(hstack([word.transform(t), char.transform(t)]).tocsr())
    return out
```

```python
# from ft.py
LLM_TEMPLATE = ("Post: {text}\n"
                "Which emotion does this post express (anger, disgust, fear, joy, sadness, or surprise)?\n"
                "Answer:")


def encode(tok, texts, max_len, kind):
    if kind == "llm":
        texts = [LLM_TEMPLATE.format(text=t) for t in texts]
    ids = tok(texts, add_special_tokens=True, truncation=False)["input_ids"]
    head = max_len // 4
    out, n_cut = [], 0
    for x in ids:
        if len(x) > max_len:  # keep head and tail (the tail holds the LLM question)
            x = x[:head] + x[-(max_len - head):]
            n_cut += 1
        out.append(x)
    return out, n_cut
```

```python
# from common.py
def scores(y_true, y_pred) -> dict:
    return {
        "macro_f1": float(f1_score(y_true, y_pred, average="macro")),
        "acc": float(accuracy_score(y_true, y_pred)),
        "weighted_f1": float(f1_score(y_true, y_pred, average="weighted")),
        "per_class_f1": dict(zip(LABELS, map(float, f1_score(y_true, y_pred, average=None, labels=range(6))))),
    }


def confusion(y_true, y_pred) -> pd.DataFrame:
    return pd.DataFrame(confusion_matrix(y_true, y_pred, labels=range(6)), index=LABELS, columns=LABELS)


# The scorer wants an `id` column (Evaluation page). sampleSubmission.csv says
# `post_id`, and a file with that header ended ERROR on 2026-11-16 (ref 58412855:
# "ID column id not found in submission"). Values are the post ids.
SUB_HEADER = ["id", "label"]


def write_submission(path: Path, test: pd.DataFrame, labels: list[str]) -> str:
    """Write id,label in sampleSubmission order and validate it."""
    sub = pd.DataFrame({"id": test["post_id"], "label": labels})
    path.parent.mkdir(parents=True, exist_ok=True)
    sub.to_csv(path, index=False, lineterminator="\n")
    return check_submission(path)


def check_submission(path: Path) -> str:
    """Raise if the file is not a valid submission; return its SHA-256."""
    with open(RAW / "sampleSubmission.csv", encoding="utf-8", newline="") as fh:
        sample = list(csv.reader(fh))
    with open(path, encoding="utf-8", newline="") as fh:
        rows = list(csv.reader(fh))
    assert sample[0] == ["post_id", "label"]
    assert rows[0] == SUB_HEADER, f"header {rows[0]} (scorer needs {SUB_HEADER})"
    assert len(rows) == len(sample) == 1441, f"{len(rows) - 1} data rows"
    assert [r[0] for r in rows[1:]] == [r[0] for r in sample[1:]], "post_id order differs from sampleSubmission"
    bad = {r[1] for r in rows[1:]} - set(LABELS)
    assert not bad, f"unknown labels {bad}"
    assert all(len(r) == 2 for r in rows), "ragged rows"
    return sha256(path)
```

```python
# from a_tfidf.py
import json

import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

ARMS = {
    "lr_c1": ("lr", 1.0),
    "lr_c4": ("lr", 4.0),
    "lr_c16": ("lr", 16.0),
    "svm_c0.1": ("svm", 0.1),
    "svm_c0.3": ("svm", 0.3),
    "svm_c1": ("svm", 1.0),
}


def model(kind, c):
    if kind == "lr":
        return LogisticRegression(C=c, max_iter=5000)
    return LinearSVC(C=c)


def run_arm(name, kind, c, train, test, folds):
    out_dir = RUNS / "a_tfidf" / name
    out_dir.mkdir(parents=True, exist_ok=True)
    y = train["y"].to_numpy()
    texts = train["text"].tolist()
    reps = {}
    for rep in [col for col in folds.columns if col.startswith("rep")]:
        oof = np.zeros((len(train), len(LABELS)))
        f = folds[rep].to_numpy()
        for k in sorted(set(f)):
            tr, va = np.where(f != k)[0], np.where(f == k)[0]
            Xtr, Xva = featurize([texts[i] for i in tr], [texts[i] for i in va])
            m = model(kind, c).fit(Xtr, y[tr])
            oof[va] = m.predict_proba(Xva) if kind == "lr" else m.decision_function(Xva)
        np.save(out_dir / f"oof_{rep}.npy", oof)
        reps[rep] = scores(y, oof.argmax(1))
    Xall, Xte = featurize(texts, test["text"].tolist())
    m = model(kind, c).fit(Xall, y)
    te = m.predict_proba(Xte) if kind == "lr" else m.decision_function(Xte)
    np.save(out_dir / "test.npy", te)
    summ = {k: {"mean": float(np.mean([r[k] for r in reps.values()])),
                "sd": float(np.std([r[k] for r in reps.values()], ddof=1))}
            for k in ("macro_f1", "acc", "weighted_f1")}
    res = {"arm": name, "kind": kind, "C": c, "reps": reps, "summary": summ,
           "test_pred_counts": {LABELS[i]: int(n) for i, n in zip(*np.unique(te.argmax(1), return_counts=True))}}
    (out_dir / "metrics.json").write_text(json.dumps(res, indent=2), encoding="utf-8")
    return res
```

```python
# from ft.py
import argparse
import contextlib
import hashlib
import json
import os
import random
import sys
import time
from pathlib import Path

import numpy as np  # noqa: E402
import torch  # noqa: E402
from transformers import (AutoConfig, AutoModelForSequenceClassification, AutoTokenizer,  # noqa: E402
                          get_cosine_schedule_with_warmup, get_linear_schedule_with_warmup)

# Yielding the GPU is normal (rota PROTOCOL §2: `done -Note yield`), so exit 0;
# the run is resumable because finished (seed, fold) pairs are on disk, and the
# missing metrics.json shows that it is incomplete.
YIELD_EXIT = 0
rota = None  # adapted: ft.py imports a local GPU-scheduler client here


def args_():
    p = argparse.ArgumentParser()
    p.add_argument("--name", required=True)
    p.add_argument("--model", required=True, help="repo id (resolved from env/models.jsonl)")
    p.add_argument("--kind", choices=["encoder", "llm"], required=True)
    p.add_argument("--rep", default="rep0")
    p.add_argument("--seeds", type=int, nargs="+", default=[0])
    p.add_argument("--folds", type=int, nargs="+", default=None)
    p.add_argument("--epochs", type=float, default=4)
    p.add_argument("--lr", type=float, default=1e-5)
    p.add_argument("--bs", type=int, default=16)
    p.add_argument("--accum", type=int, default=1)
    p.add_argument("--eval-bs", type=int, default=64)
    p.add_argument("--max-len", type=int, default=128)
    p.add_argument("--wd", type=float, default=0.01)
    p.add_argument("--warmup", type=float, default=0.1)
    p.add_argument("--sched", choices=["linear", "cosine"], default="linear")
    p.add_argument("--lora-r", type=int, default=64)
    p.add_argument("--lora-alpha", type=int, default=128)
    p.add_argument("--lora-dropout", type=float, default=0.05)
    p.add_argument("--grad-ckpt", action="store_true")
    p.add_argument("--load-4bit", action="store_true",
                   help="QLoRA: NF4 4-bit base weights (bitsandbytes), bf16 compute; for 14B LLMs on 32 GB")
    p.add_argument("--full", action="store_true",
                   help="train on all train rows (no validation), predict test; needs its own --name")
    p.add_argument("--mem-margin-gib", type=float, default=2.0)
    p.add_argument("--smoke", type=int, default=0, help="N>0: tiny random model, N train rows, CPU ok")
    p.add_argument("--path", default=None, help="smoke only: local snapshot dir (tokenizer+config) "
                                                "before models.jsonl has the record")
    p.add_argument("--device", default="cuda" if torch.cuda.is_available() else "cpu")
    return p.parse_args()


def amp(a):
    """bf16 autocast on GPU; on CPU only for the LLM (bf16 base + fp32 LoRA/head),
    so a CPU smoke exercises the same mixed-dtype path as the GPU run."""
    if a.device == "cuda":
        return torch.autocast(device_type="cuda", dtype=torch.bfloat16)
    if a.kind == "llm":
        return torch.autocast(device_type="cpu", dtype=torch.bfloat16)
    return contextlib.nullcontext()


def set_seed(s: int):
    random.seed(s)
    np.random.seed(s)
    torch.manual_seed(s)
    torch.cuda.manual_seed_all(s)


def batches(ids, idx, bs, pad_id, shuffle, gen=None):
    idx = np.asarray(idx)
    if shuffle:
        idx = idx[torch.randperm(len(idx), generator=gen).numpy()]
    else:  # length-sorted for fast eval; caller restores order via idx
        idx = idx[np.argsort([len(ids[i]) for i in idx], kind="stable")]
    for s in range(0, len(idx), bs):
        b = idx[s:s + bs]
        L = max(len(ids[i]) for i in b)
        x = torch.full((len(b), L), pad_id, dtype=torch.long)
        m = torch.zeros((len(b), L), dtype=torch.long)
        for j, i in enumerate(b):
            x[j, :len(ids[i])] = torch.tensor(ids[i])
            m[j, :len(ids[i])] = 1
        yield b, x, m


def build(a, path, tok):
    if a.smoke:
        cfg = AutoConfig.from_pretrained(path, num_labels=len(LABELS))
        small = dict(hidden_size=64, num_hidden_layers=2, num_attention_heads=2, intermediate_size=128,
                     pooler_hidden_size=64)  # DeBERTa keeps a separate pooler width
        if a.kind == "llm":
            small.update(num_key_value_heads=1, head_dim=32)
        for k, v in small.items():
            if hasattr(cfg, k):
                setattr(cfg, k, v)
        model = AutoModelForSequenceClassification.from_config(cfg)
    else:
        dtype = torch.bfloat16 if a.kind == "llm" else torch.float32
        kw = {}
        if a.kind == "llm" and a.device == "cuda":
            # load the shards straight onto the GPU: 16 GB of bf16 weights must
            # not pass through Windows RAM (only a few GB are free on this box)
            kw["device_map"] = {"": torch.cuda.current_device()}
        if a.load_4bit:
            assert a.kind == "llm" and a.device == "cuda", "--load-4bit is for LLMs on the GPU"
            from transformers import BitsAndBytesConfig
            kw["quantization_config"] = BitsAndBytesConfig(
                load_in_4bit=True, bnb_4bit_quant_type="nf4", bnb_4bit_use_double_quant=True,
                bnb_4bit_compute_dtype=torch.bfloat16)
        model = AutoModelForSequenceClassification.from_pretrained(path, num_labels=len(LABELS), dtype=dtype, **kw)
        if a.load_4bit:
            from peft import prepare_model_for_kbit_training
            model = prepare_model_for_kbit_training(model, use_gradient_checkpointing=a.grad_ckpt,
                                                    gradient_checkpointing_kwargs={"use_reentrant": False})
    model.config.pad_token_id = tok.pad_token_id
    if a.kind == "llm":
        from peft import LoraConfig, TaskType, get_peft_model
        model.config.use_cache = False
        # 4-bit: prepare_model_for_kbit_training() above already enabled checkpointing
        # and the input-grad hook; enabling again adds a second hook (R1 finding 2)
        if a.grad_ckpt and not a.load_4bit:
            model.gradient_checkpointing_enable(gradient_checkpointing_kwargs={"use_reentrant": False})
        lcfg = LoraConfig(task_type=TaskType.SEQ_CLS, r=a.lora_r, lora_alpha=a.lora_alpha,
                          lora_dropout=a.lora_dropout, target_modules="all-linear")
        model = get_peft_model(model, lcfg)
        # the classification head is trained in fp32 (modules_to_save copy)
        for n, p_ in model.named_parameters():
            if p_.requires_grad:
                p_.data = p_.data.float()
    # 4-bit weights were placed on the GPU by device_map; do not move the PEFT
    # wrapper again (R1 finding 1). Every other path is unchanged.
    return model if a.load_4bit else model.to(a.device)


def optimizer(model, a):
    decay, no_decay = [], []
    for n, p_ in model.named_parameters():
        if not p_.requires_grad:
            continue
        (no_decay if p_.ndim < 2 or "norm" in n.lower() or n.endswith(".bias") else decay).append(p_)
    return torch.optim.AdamW([{"params": decay, "weight_decay": a.wd},
                              {"params": no_decay, "weight_decay": 0.0}], lr=a.lr)


@torch.no_grad()
def predict(model, ids, idx, a, pad_id):
    model.eval()
    out = np.zeros((len(idx), len(LABELS)), dtype=np.float32)
    pos = {i: k for k, i in enumerate(idx)}
    for b, x, m in batches(ids, idx, a.eval_bs, pad_id, shuffle=False):
        with amp(a):
            logits = model(input_ids=x.to(a.device), attention_mask=m.to(a.device)).logits
        for j, i in enumerate(b):
            out[pos[i]] = logits[j].float().cpu().numpy()
    model.train()
    return out


def run_one(a, seed, fold, train, test, f, tok, path, ids_tr, ids_te, out_dir):
    tag = f"s{seed}_f{fold}"
    if (out_dir / f"oof_{tag}.npy").exists() and (out_dir / f"test_{tag}.npy").exists():
        print(f"[{tag}] done earlier, skipped", flush=True)
        return
    set_seed(seed)
    tr = np.where(f != fold)[0]
    va = np.where(f == fold)[0]
    te = np.arange(len(test))
    if a.smoke:
        tr, va, te = tr[: a.smoke], va[: max(8, a.smoke // 4)], te[:16]
    y = torch.tensor(train["y"].to_numpy())
    model = build(a, path, tok)
    opt = optimizer(model, a)
    steps_per_epoch = max(1, int(np.ceil(len(tr) / a.bs / a.accum)))
    total = max(1, int(steps_per_epoch * a.epochs))
    warm = int(a.warmup * total)
    sched = (get_cosine_schedule_with_warmup if a.sched == "cosine" else get_linear_schedule_with_warmup)(opt, warm, total)
    gen = torch.Generator().manual_seed(seed * 1000 + fold)
    hist, step, t0 = [], 0, time.time()
    va_by_ep, te_by_ep = [], []  # per-epoch predictions: the epoch count can be chosen on CV later
    n_ep = int(np.ceil(a.epochs))
    model.train()
    for ep in range(n_ep):
        micro = 0
        for b, x, m in batches(ids_tr, tr, a.bs, tok.pad_token_id, shuffle=True, gen=gen):
            with amp(a):
                logits = model(input_ids=x.to(a.device), attention_mask=m.to(a.device)).logits
            loss = torch.nn.functional.cross_entropy(logits.float(), y[b].to(a.device)) / a.accum
            loss.backward()
            micro += 1
            if micro % a.accum == 0:
                torch.nn.utils.clip_grad_norm_([p_ for p_ in model.parameters() if p_.requires_grad], 1.0)
                opt.step()
                sched.step()
                opt.zero_grad(set_to_none=True)
                step += 1
                if step >= total:
                    break
        va_logits = predict(model, ids_tr, va, a, tok.pad_token_id)
        va_by_ep.append(va_logits)
        te_by_ep.append(predict(model, ids_te, te, a, tok.pad_token_id))
        sc = scores(train["y"].to_numpy()[va], va_logits.argmax(1))
        hist.append({"epoch": ep + 1, "step": step, "sec": round(time.time() - t0, 1),
                     "val_macro_f1": sc["macro_f1"], "val_acc": sc["acc"]})
        print(f"[{tag}] ep{ep + 1} step {step}/{total} macroF1 {sc['macro_f1']:.4f} acc {sc['acc']:.4f} "
              f"{time.time() - t0:.0f}s", flush=True)
        if step >= total:
            break
        if rota is not None and a.device == "cuda" and rota.yield_requested():
            print(f"[{tag}] yield requested: dropping this unfinished fold and exiting", flush=True)
            sys.exit(YIELD_EXIT)
    te_logits = te_by_ep[-1]
    # [epoch, row, class]; rows are this fold's val rows (va) and the test rows (te)
    np.save(out_dir / f"ep_val_{tag}.npy", np.stack(va_by_ep))
    np.save(out_dir / f"ep_test_{tag}.npy", np.stack(te_by_ep))
    np.save(out_dir / f"ep_validx_{tag}.npy", va)
    oof = np.full((len(train), len(LABELS)), np.nan, dtype=np.float32)
    oof[va] = va_logits
    tst = np.full((len(test), len(LABELS)), np.nan, dtype=np.float32)
    tst[te] = te_logits
    np.save(out_dir / f"test_{tag}.npy", tst)
    np.save(out_dir / f"oof_{tag}.npy", oof)  # written last: its presence marks the pair complete
    peak = torch.cuda.max_memory_reserved() / 2**30 if a.device == "cuda" else None
    (out_dir / f"hist_{tag}.json").write_text(json.dumps({"hist": hist, "peak_reserved_gib": peak,
                                                           "n_train": len(tr), "n_val": len(va)}, indent=1))
    del model, opt
    if a.device == "cuda":
        torch.cuda.empty_cache()
        torch.cuda.reset_peak_memory_stats()
    if rota is not None and a.device == "cuda" and rota.yield_requested():
        print(f"[{tag}] yield requested after a finished fold: exiting", flush=True)
        sys.exit(YIELD_EXIT)


FULL_GEN_OFFSET = 100  # shuffle generator seed*1000+offset; fold runs use seed*1000+fold (0-4)
CFG_KEYS = ("model", "kind", "epochs", "lr", "bs", "accum", "max_len", "wd", "warmup", "sched",
            "lora_r", "lora_alpha", "lora_dropout", "grad_ckpt", "load_4bit", "smoke")


def cfg_hash(a, rev: str) -> str:
    """Everything that changes a full-data model; a .done marker from another config is never reused."""
    blob = json.dumps({**{k: getattr(a, k) for k in CFG_KEYS}, "revision": rev}, sort_keys=True)
    return hashlib.sha256(blob.encode("utf-8")).hexdigest()


def run_full(a, seed, train, test, tok, path, rev, ids_tr, ids_te, out_dir):
    """Train on every train row with the CV recipe (no validation) and predict test.

    Same loop as run_one minus the per-epoch validation; the epoch count and all
    hyperparameters come from the CV runs. full_s{seed}.done is written last
    (atomically) and records the config hash and the SHA-256 of the test array.
    """
    tag = f"full_s{seed}"
    done = out_dir / f"{tag}.done"
    npy = out_dir / f"full_test_s{seed}.npy"
    ch = cfg_hash(a, rev)
    if done.exists():
        rec = json.loads(done.read_text(encoding="utf-8"))
        if rec.get("cfg") != ch:
            raise SystemExit(f"[{tag}] {done.name} was written by another config; use a new --name")
        if npy.exists() and np.load(npy).shape == (len(test), len(LABELS)):
            print(f"[{tag}] done earlier, skipped", flush=True)
            return
        raise SystemExit(f"[{tag}] {done.name} exists but {npy.name} is missing or has the wrong shape")
    set_seed(seed)
    tr = np.arange(len(train))
    te = np.arange(len(test))
    if a.smoke:
        tr, te = tr[: a.smoke], te[:16]
    y = torch.tensor(train["y"].to_numpy())
    model = build(a, path, tok)
    opt = optimizer(model, a)
    steps_per_epoch = max(1, int(np.ceil(len(tr) / a.bs / a.accum)))
    total = max(1, int(steps_per_epoch * a.epochs))
    warm = int(a.warmup * total)
    sched = (get_cosine_schedule_with_warmup if a.sched == "cosine" else get_linear_schedule_with_warmup)(opt, warm, total)
    gen = torch.Generator().manual_seed(seed * 1000 + FULL_GEN_OFFSET)
    print(f"[{tag}] full-data training: {len(tr)} rows, {total} optimizer steps", flush=True)
    hist, step, t0 = [], 0, time.time()
    n_ep = int(np.ceil(a.epochs))
    model.train()
    for ep in range(n_ep):
        micro = 0
        loss_acc = torch.zeros((), device=a.device)  # summed on the device: no per-step sync
        for b, x, m in batches(ids_tr, tr, a.bs, tok.pad_token_id, shuffle=True, gen=gen):
            with amp(a):
                logits = model(input_ids=x.to(a.device), attention_mask=m.to(a.device)).logits
            loss = torch.nn.functional.cross_entropy(logits.float(), y[b].to(a.device)) / a.accum
            loss.backward()
            loss_acc += loss.detach() * a.accum
            micro += 1
            if micro % a.accum == 0:
                torch.nn.utils.clip_grad_norm_([p_ for p_ in model.parameters() if p_.requires_grad], 1.0)
                opt.step()
                sched.step()
                opt.zero_grad(set_to_none=True)
                step += 1
                if step >= total:
                    break
        mean_loss = float(loss_acc) / max(1, micro)
        hist.append({"epoch": ep + 1, "step": step, "sec": round(time.time() - t0, 1), "train_loss": round(mean_loss, 5)})
        print(f"[{tag}] ep{ep + 1} step {step}/{total} train_loss {mean_loss:.4f} {time.time() - t0:.0f}s", flush=True)
        if step >= total:
            break
        if rota is not None and a.device == "cuda" and rota.yield_requested():
            print(f"[{tag}] yield requested: dropping this unfinished seed and exiting", flush=True)
            sys.exit(YIELD_EXIT)
    tst = np.full((len(test), len(LABELS)), np.nan, dtype=np.float32)
    tst[te] = predict(model, ids_te, te, a, tok.pad_token_id)
    np.save(npy, tst)
    peak = torch.cuda.max_memory_reserved() / 2**30 if a.device == "cuda" else None
    rec = {"cfg": ch, "seed": seed, "n_train": len(tr), "n_test_pred": len(te), "steps": step,
           "total_steps": total, "hist": hist, "peak_reserved_gib": peak,
           "sha256": hashlib.sha256(npy.read_bytes()).hexdigest(),
           "finished": time.strftime("%Y-%m-%d %H:%M:%S")}
    tmp = out_dir / f"{tag}.done.tmp"
    tmp.write_text(json.dumps(rec, indent=1), encoding="utf-8")
    os.replace(tmp, done)  # the marker appears only after the array is complete
    del model, opt
    if a.device == "cuda":
        torch.cuda.empty_cache()
        torch.cuda.reset_peak_memory_stats()
    if rota is not None and a.device == "cuda" and rota.yield_requested():
        print(f"[{tag}] yield requested after a finished seed: exiting", flush=True)
        sys.exit(YIELD_EXIT)


def softmax(z):
    z = z - np.nanmax(z, axis=1, keepdims=True)
    e = np.exp(z)
    return e / e.sum(axis=1, keepdims=True)


def aggregate(a, train, out_dir):
    """Aggregate every seed found on disk (a later invocation may add seeds to
    the same run dir; aggregating only a.seeds would drop the earlier ones)."""
    import re
    y = train["y"].to_numpy()
    seeds_on_disk = sorted({int(re.search(r"oof_s(\d+)_f", p.name).group(1)) for p in out_dir.glob("oof_s*_f*.npy")})
    res = {"name": a.name, "args": vars(a), "seeds_on_disk": seeds_on_disk, "seeds": {}}
    seed_probs = []
    for seed in seeds_on_disk:
        parts = sorted(out_dir.glob(f"oof_s{seed}_f*.npy"))
        if not parts:
            continue
        oof = np.nanmean(np.stack([np.load(p) for p in parts]), axis=0) if len(parts) > 1 else np.load(parts[0])
        have = ~np.isnan(oof).any(1)
        pr = np.full_like(oof, np.nan)
        pr[have] = softmax(oof[have])
        seed_probs.append(pr)
        res["seeds"][seed] = {"n_folds": len(parts), "n_rows": int(have.sum()), **scores(y[have], pr[have].argmax(1))}
    if seed_probs:
        mean = np.nanmean(np.stack(seed_probs), axis=0)
        have = ~np.isnan(mean).any(1)
        res["seed_mean"] = {"n_rows": int(have.sum()), **scores(y[have], mean[have].argmax(1))}
        np.save(out_dir / f"oof_probs_{a.rep}.npy", mean)
        sm = [s["macro_f1"] for s in res["seeds"].values()]
        res["single_seed_macro_f1"] = {"mean": float(np.mean(sm)), "sd": float(np.std(sm, ddof=1)) if len(sm) > 1 else None}
    tests = [np.load(p) for p in sorted(out_dir.glob("test_s*_f*.npy"))]
    if tests:
        tp = np.stack([softmax(np.nan_to_num(t, nan=0.0)) for t in tests])
        np.save(out_dir / "test_probs.npy", tp.mean(0))
        res["n_test_models"] = len(tests)
    (out_dir / "metrics.json").write_text(json.dumps(res, indent=2), encoding="utf-8")
    return res
```

```python
# from blend.py
import argparse
import itertools
import json

import numpy as np
from sklearn.metrics import f1_score


def load_run(name: str, rep: str):
    d = RUNS / name
    if (d / f"oof_probs_{rep}.npy").exists():  # ft.py runs (already softmax, seed-averaged)
        oof, te = np.load(d / f"oof_probs_{rep}.npy"), np.load(d / "test_probs.npy")
    else:  # a_tfidf arms: LR probabilities or SVM decision scores
        oof, te = np.load(d / f"oof_{rep}.npy"), np.load(d / "test.npy")
        if not np.allclose(oof.sum(1), 1, atol=1e-3):
            oof, te = softmax(oof), softmax(te)
    return oof, te


def macro(y, pred):
    return f1_score(y, pred, average="macro")


def fit_blend_weights(ps, y, grid=np.linspace(0, 1, 11)):
    """Grid over the simplex (2 or 3 models); equal weights beyond that."""
    if len(ps) == 1:
        return np.array([1.0])
    if len(ps) > 3:
        return np.ones(len(ps)) / len(ps)
    best = (-1.0, None)
    for ws in itertools.product(grid, repeat=len(ps) - 1):
        if sum(ws) > 1 + 1e-9:
            continue
        w = np.array(list(ws) + [1 - sum(ws)])
        f = macro(y, sum(wi * p for wi, p in zip(w, ps)).argmax(1))
        if f > best[0]:
            best = (f, w)
    return best[1]


# adapted: blend.py's own softmax() is ft.py's with max for nanmax, the same on rows without NaN;
# load_run() uses the ft.py copy above
```

```python
# adapted
SUBMISSIONS = {
    "s04": {"file": "s04_c_qwen3-8b_lora_e3_s0.csv", "members": [["C", [0]]], "weights": None,
            "sha256": "7cb1562ac3e3891a3ae38efa4715ab5d5982f10dc9c65e242f3c0094b0b00be6"},
    "s05": {"file": "s05_blend_B0.4_C0.6.csv", "members": [["B", [0, 1]], ["C", [0]]], "weights": [0.4, 0.6],
            "sha256": "b969bb2468b105bbe401926f66d341d2c7269ab1695c665202ae2b69ba98d2a9"},
    "s07": {"file": "s07_blend_B3s_C2s_Cq25_2s_eq.csv", "members": [["B", [0, 1, 2]], ["C", [0, 1]], ["C'", [0, 1]]], "weights": None,
            "sha256": "3cb9afacefb06c9837e4b59826883da0c4f6c8d93db7528eb47b39969e2c395d"},
}


def member_arrays(key):
    """Logits of one member for every saved (seed, fold): from competition/preds/<run>.npz (replay)
    or from the run directory that train_member() wrote (retrain)."""
    run = MEMBERS[key]["run"]
    if LAB2_MODE == "replay":
        with np.load(PREDS / f"{run}.npz") as z:
            return {k: z[k] for k in z.files}
    return {p.stem: np.load(p) for p in sorted((RUNS / run).glob("*.npy"))
            if re.fullmatch(r"(oof|test)_s\d+_f\d+|full_test_s\d+", p.stem)}


def member_probs(key, seeds):
    """(OOF, test) probabilities of one member from `seeds` only: aggregate() on a directory that holds
    just those seeds, as the run directory did when the submission was written, then load_run()."""
    arrays = member_arrays(key)
    if any(k.startswith("full_test_") for k in arrays):  # full-data run: no OOF; aggregate()'s test step
        tp = np.stack([softmax(np.nan_to_num(arrays[f"full_test_s{s}"], nan=0.0)) for s in seeds])
        return None, tp.mean(0)
    name = f"{MEMBERS[key]['run']}@s{'-'.join(map(str, seeds))}"
    (RUNS / name).mkdir(parents=True, exist_ok=True)
    for k, v in arrays.items():
        if int(re.search(r"_s(\d+)_", k).group(1)) in seeds:
            np.save(RUNS / name / f"{k}.npy", v)
    aggregate(argparse.Namespace(name=name, rep="rep0"), train, RUNS / name)
    return load_run(name, "rep0")


def build_submission(sid):
    """blend.py's main() for one row of SUBMISSIONS: weighted sum of the members' probabilities, argmax,
    write_submission(). Returns the file's SHA-256, the blended OOF probabilities and the test labels."""
    spec = SUBMISSIONS[sid]
    loaded = [member_probs(k, s) for k, s in spec["members"]]
    w_all = np.array(spec["weights"]) / sum(spec["weights"]) if spec["weights"] else np.ones(len(loaded)) / len(loaded)
    oof = None
    if all(o is not None for o, _ in loaded):
        have = np.all([~np.isnan(o).any(1) for o, _ in loaded], axis=0)
        oofs = [o[have] for o, _ in loaded]
        oof = np.full((len(have), len(LABELS)), np.nan)
        oof[have] = sum(wi * o for wi, o in zip(w_all, oofs))
    tes = [t for _, t in loaded]
    cw = np.ones(len(LABELS))
    pt = sum(wi * t for wi, t in zip(w_all, tes)) * cw
    labels = [LABELS[i] for i in pt.argmax(1)]
    return write_submission(SUBS / spec["file"], test, labels), oof, pt.argmax(1)
```
